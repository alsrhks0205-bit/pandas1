## 4. loc vs iloc — 이름표 vs 위치
같은 0:2 범위를 loc과 iloc으로 선택해 끝을 포함하는지 비교합니다.

```python
tips.loc[1 , "tip"]
```

```python
tips.iloc[1 , 1]
```

```python
tips.iloc[1:4 , 0:2]
```

## 5. loc — 이름표로 행·열 선택
loc으로 행 하나, 행 범위와 열 목록, 열 이름 범위를 차례로 선택합니다.

```python
x = ["total_bill", "tip", "day"]
print(tips.loc[3])                           # 인덱스 3인 행 1개 → Series
print(tips.loc[0:3, x])   # 행 범위 + 열 목록
print(tips.loc[:, "sex":"day"].head(2))      # 모든 행 + sex~day 열 범위
```

## 6. iloc — 위치 번호로 행·열 선택
iloc으로 위치 범위, 마지막 행, 원하는 위치만 골라 선택합니다.

```python
print(tips.iloc[0:3, 0:2])                   # 0~2번째 행, 0~1번째 열
print(tips.iloc[-1])                         # 마지막 행
print(tips.iloc[[0, 10, 20], [1, 4]])        # 원하는 위치만 골라서
```

## 7. 조건 필터링 — True/False 마스크
팁 5 이상 조건으로 마스크를 만들고 True인 행만 남깁니다.

```python
mask = tips["tip"] >= 5
tips[mask]
```

## 8. 복합 조건 — & · | · ~
그리고·또는·아님 조건으로 행을 걸러 각각 몇 행이 남는지 확인합니다.

```python
# 일요일(Sun)에 온 손님중에서 팁 5달러 이상 준 사람
mask = (tips["day"] == "Sun") & (tips["tip"] >= 5)
tips[mask]
```

```python
# 주말에 온 손님
mask = (tips["day"] == "Sun") | (tips["day"] == "Sat")
tips[mask]
```

```python
# 흡연자가 아닌 손님
mask = ~(tips["smoker"] == "Yes")
tips[mask]
```

## 9. 목록·범위 조건 — isin · between
isin으로 주말 행을, between으로 결제 금액 20에서 30 사이 행을 골라 확인합니다.

```python
# 주말 손님
mask = tips["day"].isin( ["Sat" , "Sun"] )
tips[mask]
```

```python
# 결제금액 20 ~ 30
mask = tips["total_bill"].between(20 , 30)
tips[mask]
```

## 10. loc[조건, 열] — 행 필터와 열 선택을 한 번에
점심·4인 이상 조건에 맞는 행에서 필요한 열 4개만 선택합니다.

```python
mask = (tips["time"] == "Lunch") & (tips["size"] >= 4)     # 점심 + 4인 이상
lunch_group = tips.loc[mask, ["total_bill", "tip", "size", "day"]]  # 행 조건 + 열 선택
print(lunch_group)
```

## 11. query — 조건을 문장처럼
query 문자열로 조건을 쓰고, @변수와 in 연산자도 사용해 봅니다.

```python
# 일요일 + 팁5달러이상
tips.query("tip >= 5 and day == 'Sun'")
```

```python
# 팁을 7달러 이상 준 사람
x = 7
tips.query("tip >= @x") # 변수를 취급할 땐, "@" 기호 사용
```

```python
tips.query('day in ["Sat","Sun"] and size > 4')
```

## 12. 정렬 — sort_values (한 열)
팁 기준 내림차순으로 정렬해 상위 3행을 확인합니다.

```python
tips.sort_values(by="tip" , ascending=False).head(3)
```

## 13. 정렬 — 여러 기준
요일은 오름차순, 같은 요일 안에서는 팁 내림차순으로 정렬합니다.

```python
tips.sort_values(
    by=["day" , "tip"] , 
    ascending=[True,False]
)
```
